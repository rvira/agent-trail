# AgentTrail — Week 1 Pilot Exploration

This notebook explores the staged `data/pilot_repos.parquet` dataset using **DuckDB**, **Pandas**, and **Seaborn**.

```python
import duckdb
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

con = duckdb.connect()
df = con.execute("SELECT * FROM '../data/pilot_repos.parquet'").fetchdf()
df.head()
```

```python
con.execute("""
    SELECT
        language,
        COUNT(*) AS repos,
        SUM(adopted) AS adopted_repos,
        ROUND(AVG(stars), 1) AS mean_stars
    FROM '../data/pilot_repos.parquet'
    GROUP BY language
    ORDER BY repos DESC
""").fetchdf()
```
